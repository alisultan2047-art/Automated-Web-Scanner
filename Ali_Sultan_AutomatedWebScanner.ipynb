```python
import requests

def run_web_scanner(url):
    """
    Automated Web Scanner script to check security response headers
    and CORS policies for a target website.
    """
    print("=" * 60)
    print(f"STARTING SECURITY SCAN FOR: {url}")
    print("=" * 60 + "\n")

    try:
        # Step 1: Send HTTP request to target URL
        response = requests.get(url, timeout=10)
        headers = response.headers

        print(f"[+] HTTP Status Code: {response.status_code}\n")
        print("--- AUDITING SECURITY HEADERS ---")

        # Step 2: Define important security headers to audit
        security_headers = {
            'Strict-Transport-Security': 'Enforces HTTPS encryption.',
            'Content-Security-Policy': 'Prevents Cross-Site Scripting (XSS) and code injection.',
            'X-Frame-Options': 'Protects against Clickjacking attacks.',
            'X-Content-Type-Options': 'Prevents MIME-type sniffing vulnerabilities.'
        }

        # Step 3: Check presence of each header
        for header, description in security_headers.items():
            if header in headers:
                print(f"[PASSED] {header}")
                print(f"         Value: {headers[header]}")
            else:
                print(f"[FAILED] {header} is MISSING.")
                print(f"         Risk: Website lacks standard protection ({description})")
            print("-" * 50)

        # Step 4: Audit CORS (Cross-Origin Resource Sharing) configuration
        print("\n--- AUDITING CORS POLICY ---")
        if 'Access-Control-Allow-Origin' in headers:
            cors_value = headers['Access-Control-Allow-Origin']
            print(f"[+] CORS Header Found: {cors_value}")
            if cors_value == '*':
                print("[!] VULNERABILITY WARNING: Wildcard '*' allows any external domain to access resources.")
            else:
                print("[+] CORS policy is restricted to specific origin(s).")
        else:
            print("[*] No 'Access-Control-Allow-Origin' header explicitly declared.")

    except Exception as e:
        print(f"[!] Error executing scan: {e}")

    print("\n" + "=" * 60)
    print("SCAN COMPLETE")
    print("=" * 60)

# Run scan on test domain
run_web_scanner("https://example.com")
```
